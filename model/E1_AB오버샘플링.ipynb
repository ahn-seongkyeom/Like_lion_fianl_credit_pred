```python
import pandas as pd
from sklearn.utils import resample
```

```python
df = pd.read_parquet("merge/result/Segment_merge_ver_03.parquet")
```

```python
# 각 Segment 별 데이터 분리
df_E = df[df['Segment'] == 'E']
df_D = df[df['Segment'] == 'D']
df_C = df[df['Segment'] == 'C']
df_A = df[df['Segment'] == 'A']
df_B = df[df['Segment'] == 'B']
```

```python
# A, B 오버샘플링 
df_A_upsampled = resample(df_A,
                          replace=True,      # 복원추출
                          n_samples=10000,   # 원하는 수로 조절
                          random_state=42)

df_B_upsampled = resample(df_B,
                          replace=True,
                          n_samples=10000,
                          random_state=42)
```

```python
# 오버샘플링된 데이터 + 원본 C, D, E 병합
df_oversampled = pd.concat([df_E, df_D, df_C, df_A_upsampled, df_B_upsampled])

# 클래스 섞기 (셔플)
df_oversampled = df_oversampled.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
print(df_oversampled['Segment'].value_counts())
```

```python
# 새로운 파일로 저장
df_oversampled.to_csv("merge/result/E1_AB오버샘플링.csv", index=False)
```
