```python
import pandas as pd
import os, re
from google.colab import drive
drive.mount('/content/drive')
```

```python
# 데이터 경로
print(os.getcwd())
path = '/content/drive/MyDrive/Colab_Notebooks'
prep_folder = os.listdir(path)
```

```python
score_list = [p for p in prep_folder if re.search(r'.*est_(?!1000).*score.*\.csv$', p)]
score_list
```

```python
len(score_list)
```

```python
score_df = pd.DataFrame()

for score in score_list:
    df = pd.read_csv(os.path.join(path, score))
    df.index = [score]  # 파일명을 행 인덱스로 지정
    score_df = pd.concat([score_df, df])
```

```python
score_df.to_csv(os.path.join(path, 'score_result.csv'))
```
